# Setup

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

os.chdir(r"D:\repos\analysis-for-action")

# Load the dataset
vuln = pd.read_csv("learning_resources/data/vulnerable.csv")
```

# Prepare your data

```python
vuln_2007 = vuln[(vuln['year'] == 2007) & (vuln['country'].isin(["Mauritius", "Denmark", "Canada", "Netherlands"]))]
# order from largest to smallest
vuln_2007 = vuln_2007.sort_values(by='vulnerable_pop', ascending=False)

print(vuln_2007)
```

# Make the base pie chart

```python
vuln_pie = plt.pie(vuln_2007['vulnerable_pop'], labels=vuln_2007['country'], startangle=90, counterclock=False)
# Equal aspect ratio ensures that pie is drawn as a circle.
plt.axis('equal')
# Set label alignment to horizontal
plt.setp(vuln_pie[1], horizontalalignment='center')
# Change alignment based on the position of the label
for label in vuln_pie[1]:
    x, y = label.get_position()

    label.set_position((x, y))

    if x > 0.5:
        label.set_horizontalalignment('left')
    elif x < -0.5:
        label.set_horizontalalignment('right')
    else:
        label.set_horizontalalignment('center')
```

# Add white border between sectors

```python
vuln_pie_white_border = plt.pie(vuln_2007['vulnerable_pop'], labels=vuln_2007['country'], startangle=90, counterclock=False,
                                # add white border between sectors
                                wedgeprops={'edgecolor': 'white'})
# Equal aspect ratio ensures that pie is drawn as a circle
plt.axis('equal')  
# Change alignment based on the position of the label
for label in vuln_pie_white_border[1]:
    x, y = label.get_position()

    label.set_position((x, y))

    if x > 0.5:
        label.set_horizontalalignment('left')
    elif x < -0.5:
        label.set_horizontalalignment('right')
    else:
        label.set_horizontalalignment('center')
```

# Change sector colours to ensure high contrast

```python
country_colours = {
    "Canada": "#D17E38",
    "Netherlands": "#444443",
    "Denmark": "#3EA27F",
    "Mauritius": "#752111"}

vuln_pie_colours = plt.pie(vuln_2007['vulnerable_pop'], labels=vuln_2007['country'], startangle=90, counterclock=False,
                                # add white border between sectors
                                wedgeprops={'edgecolor': 'white'},
                                colors=[country_colours[country] for country in vuln_2007['country']])
# Equal aspect ratio ensures that pie is drawn as a circle
plt.axis('equal')  
# Change alignment based on the position of the label
for label in vuln_pie_colours[1]:
    x, y = label.get_position()

    label.set_position((x, y))

    if x > 0.5:
        label.set_horizontalalignment('left')
    elif x < -0.5:
        label.set_horizontalalignment('right')
    else:
        label.set_horizontalalignment('center')
```

# Add percentages as labels

```python
country_colours = {
    "Canada": "#D17E38",
    "Netherlands": "#444443",
    "Denmark": "#3EA27F",
    "Mauritius": "#752111"}

percentages = (vuln_2007['vulnerable_pop'] / vuln_2007['vulnerable_pop'].sum() * 100)

country_labels = [f"{country} [{percentage:.1f}%]" for country, percentage in zip(vuln_2007['country'], percentages)]

vuln_pie_percentage = plt.pie(vuln_2007['vulnerable_pop'], labels=country_labels, startangle=90, counterclock=False,
                                # add white border between sectors
                                wedgeprops={'edgecolor': 'white'},
                                colors=[country_colours[country] for country in vuln_2007['country']])
# Equal aspect ratio ensures that pie is drawn as a circle
plt.axis('equal')  
# Change alignment based on the position of the label (adjusted to align Denmark to right due to increased label length)
for label in vuln_pie_percentage[1]:
    x, y = label.get_position()

    label.set_position((x, y))

    if x > 0.1:
        label.set_horizontalalignment('left')
    elif x < -0.1:
        label.set_horizontalalignment('right')
    else:
        label.set_horizontalalignment('center')
```

# Exporting (.svg file type)

```python
country_colours = {
    "Canada": "#D17E38",
    "Netherlands": "#444443",
    "Denmark": "#3EA27F",
    "Mauritius": "#752111"}

percentages = (vuln_2007['vulnerable_pop'] / vuln_2007['vulnerable_pop'].sum() * 100)

country_labels = [f"{country} [{percentage:.1f}%]" for country, percentage in zip(vuln_2007['country'], percentages)]

vuln_pie_percentage = plt.pie(vuln_2007['vulnerable_pop'], labels=country_labels, startangle=90, counterclock=False,
                                # add white border between sectors
                                wedgeprops={'edgecolor': 'white'},
                                colors=[country_colours[country] for country in vuln_2007['country']])
# Equal aspect ratio ensures that pie is drawn as a circle
plt.axis('equal')  
# Change alignment based on the position of the label (adjusted to align Denmark to right due to increased label length)
for label in vuln_pie_percentage[1]:
    x, y = label.get_position()

    label.set_position((x, y))

    if x > 0.1:
        label.set_horizontalalignment('left')
    elif x < -0.1:
        label.set_horizontalalignment('right')
    else:
        label.set_horizontalalignment('center')

# save the pie chart as an svg file
plt.savefig("learning_resources/outputs_and_reporting/data_visualisation/basic_data_visualisation/python/example_outputs/vuln_pie.svg", format="svg")
```

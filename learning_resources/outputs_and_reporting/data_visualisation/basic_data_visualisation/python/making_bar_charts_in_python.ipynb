# Setup

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

os.chdir(r"D:\repos\analysis-for-action")

# Load the dataset
vuln = pd.read_csv("learning_resources/data/vulnerable.csv")
```

# Prepare your data

```python
# Filter the dataset to 2007 and display the top 6 countries with the highest vulnerable population
vuln_2007_top_6 = (
    vuln.loc[vuln["year"] == 2007].nlargest(6, "vulnerable_pop")
)

vuln_2007_top_6
```

# Clustered bar chart

```python
vuln_2002_top_6 = (
    # Select the 6 countries with the highest vulnerable population in 2002
    vuln.loc[vuln["year"].isin([2002])].nlargest(6, "vulnerable_pop")
)

vuln_2007_top_6 = (
    # Select the 6 countries with the highest vulnerable population in 2007
    vuln.loc[vuln["year"].isin([2007])].nlargest(6, "vulnerable_pop")
)

vuln_2002_and_2007_top_6 = pd.concat([vuln_2002_top_6, vuln_2007_top_6])

vuln_2002_and_2007_top_6
```

```python
# Set the theme for the plot
sns.set_theme(style="white")

vuln_cluster = sns.barplot(
    data = vuln_2002_and_2007_top_6,
    x = "vulnerable_pop",
    y = "country",
    # Change colour of bars based on year
    hue = "year",
    palette = {2002: "#48553f", 2007: "#f59342"}
)
# Set the gridlines for the x-axis
vuln_cluster.xaxis.grid(True, linestyle='-', linewidth=0.5, color='lightgrey')
# Remove scientific notation
vuln_cluster.ticklabel_format(axis='x', style='plain')
# Set y-axis title to be horizontal
vuln_cluster.yaxis.label.set_rotation(0)
vuln_cluster.yaxis.label.set_horizontalalignment('right')
vuln_cluster.yaxis.label.set_verticalalignment('center')
vuln_cluster.yaxis.set_label_coords(-0.1, 1.02)
```

# Stacked bar chart

```python
# Get cumulative vulnerable population for each country across all years
top_countries = (
    vuln.groupby("country")["vulnerable_pop"].sum()
)
top_6 = top_countries.nlargest(6).index
print(top_6)

vuln_top_6 = vuln[
vuln["country"].isin(top_6)
]

vuln_top_6
```

```python
data = (
    vuln_top_6
    .pivot_table(
        index = "year",
        columns = "country",
        values = "vulnerable_pop",
        aggfunc = "sum",
        observed = False
    )
)


# Set the theme for the plot
sns.set_theme(style="white")

country_colours = {
"Japan": "#1E2A1F",
"Brazil": "#D17E38",
"Indonesia": "#444443",
"United States": "#3EA27F",
"India": "#752111",
"China": "#618DD3"
}

# Reorder the columns to match the order of the top 6 countries
data = data[top_6]

# Set the gridlines for the x-axis
vuln_stacked_bar = data.plot(
    kind = "bar",
    stacked = True,
    color = country_colours
)
# Set the gridlines for the y-axis
vuln_stacked_bar.yaxis.grid(True, linestyle='-', linewidth=0.5, color='lightgrey')
# Add axis titles
vuln_stacked_bar.set_xlabel("Year")
vuln_stacked_bar.set_ylabel("Vulnerable Population")
# Remove scientific notation
vuln_stacked_bar.ticklabel_format(axis='y', style='plain')
# Set y-axis title to be horizontal
vuln_stacked_bar.yaxis.label.set_rotation(0)
vuln_stacked_bar.yaxis.label.set_horizontalalignment('right')
vuln_stacked_bar.yaxis.label.set_verticalalignment('center')
vuln_stacked_bar.yaxis.set_label_coords(-0.1, 1.02)
# Position the legend outside the plot
vuln_stacked_bar.legend(
title="Country",
bbox_to_anchor=(1.02, 1),
loc="upper left"
)

plt.show()
```

# 100% Stacked bar chart

```python
# Prepare the dataset for a horizontal stacked bar chart

vuln_1997_2002_2007_percent = vuln.loc[
    vuln["year"].isin([1997, 2002, 2007]) &
    vuln["country"].isin(["Afghanistan", "Canada", "Iraq", "Peru", "Romania"])
]

data = (
    vuln_1997_2002_2007_percent
    .pivot_table(
        index = "year",
        columns = "country",
        values = "vulnerable_pop",
        aggfunc = "sum",
        observed = False
    )
)
```

```python
# Create 100% stacked bar chart to show the percentage of vulnerable population for each country in 1997, 2002 and 2007

vuln_stacked_bar_percent = data.div(data.sum(axis=1), axis=0).plot(
    kind = "barh",
    stacked = True,
    color = {
        "Afghanistan": "#1E2A1F",
        "Canada": "#D17E38",
        "Iraq": "#444443",
        "Peru": "#3EA27F",
        "Romania": "#752111"
    }
)

# Add axis titles
vuln_stacked_bar_percent.set_xlabel("Year")

# Set x-axis to be in range 0-100% and ticks to be at intervals of 20%
vuln_stacked_bar_percent.set_xlim(0, 1)
vuln_stacked_bar_percent.set_xticks([0, 0.2, 0.4, 0.6, 0.8, 1])

# Position the legend outside the plot
vuln_stacked_bar_percent.legend(
title="Country",
bbox_to_anchor=(1.2, 1),
loc="upper left"
)

# Set y-axis title to be horizontal
vuln_stacked_bar_percent.yaxis.label.set_rotation(0)
vuln_stacked_bar_percent.yaxis.label.set_horizontalalignment('right')
vuln_stacked_bar_percent.yaxis.label.set_verticalalignment('center')
vuln_stacked_bar_percent.yaxis.set_label_coords(-0.1, 1.02)

# Create labels at the end of bars showing the total population for each year
for i, total in enumerate(data.sum(axis=1)):
    vuln_stacked_bar_percent.text(
        1.01,
        i,
        f"{total:,}",
        va="center",
        ha="left"
    )

# Add individual labels for each country showing the vulnerable population of each country for each year
for j, (_, row) in enumerate(data.iterrows()):
    row_percent = row / row.sum()
    cumulative = 0
    for country, value in row.items():
        width = row_percent[country]
        vuln_stacked_bar_percent.text(
            cumulative + width / 2,
            j,
            f"{value:,}",
            va="center",
            ha="center",
            color="white"
        )
        cumulative += width
```

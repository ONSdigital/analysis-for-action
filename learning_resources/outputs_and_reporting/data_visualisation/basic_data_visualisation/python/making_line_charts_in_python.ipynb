# Setup

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

os.chdir(r"D:\repos\analysis-for-action")

# Load the dataset
vuln = pd.read_csv("learning_resources/data/vulnerable.csv")
```

# Prepare your data

```python
# Prepare your data to make it suitable for a line chart

vuln_4_country = vuln[vuln["country"].isin(["United Kingdom", "Malawi", "Nepal", "Argentina"])]
```

# Make the base bar chart

```python
# Set country colours
country_colours = {
    "United Kingdom": "#1E2A16",
    "Argentina": "#D17E38",
    "Malawi": "#444443",
    "Nepal": "#3EA27F"
}

vuln_line = sns.lineplot(
    data=vuln_4_country,
    x="year",
    y="vulnerable_pop",
    hue="country",
    palette=country_colours
)
# Set the gridlines for the x-axis
vuln_line.yaxis.grid(True, linestyle='-', linewidth=0.5, color='lightgrey')
# Set y-axis title to be horizontal
vuln_line.yaxis.label.set_rotation(0)
vuln_line.yaxis.label.set_horizontalalignment('right')
vuln_line.yaxis.label.set_verticalalignment('center')
vuln_line.yaxis.set_label_coords(-0.1, 1.02)
# Remove scientific notation from y-axis
vuln_line.ticklabel_format(axis='y', style='plain')
# Add labels to the end of the lines for each country
for country, df_country in vuln_4_country.groupby("country"):
    vuln_line.text(
        x=df_country["year"].max(),
        y=df_country["vulnerable_pop"].iloc[-1] + 200000,
        s=country,
        fontsize=10,
        va='center',
        ha='right'
    )
# Remove legend
vuln_line.legend_.remove()
```

# Add axis titles

```python
# Set country colours
country_colours = {
    "United Kingdom": "#1E2A16",
    "Argentina": "#D17E38",
    "Malawi": "#444443",
    "Nepal": "#3EA27F"
}

vuln_line_titles = sns.lineplot(
    data=vuln_4_country,
    x="year",
    y="vulnerable_pop",
    hue="country",
    palette=country_colours
)
# Set axis titles
vuln_line_titles.set_xlabel("Year")
vuln_line_titles.set_ylabel("Vulnerable Population")
# Set the gridlines for the x-axis
vuln_line_titles.yaxis.grid(True, linestyle='-', linewidth=0.5, color='lightgrey')
# Set y-axis title to be horizontal
vuln_line_titles.yaxis.label.set_rotation(0)
vuln_line_titles.yaxis.label.set_horizontalalignment('right')
vuln_line_titles.yaxis.label.set_verticalalignment('center')
vuln_line_titles.yaxis.set_label_coords(-0.1, 1.02)
# Remove scientific notation from y-axis
vuln_line_titles.ticklabel_format(axis='y', style='plain')
# Add labels to the end of the lines for each country
for country, df_country in vuln_4_country.groupby("country"):
    vuln_line_titles.text(
        x=df_country["year"].max(),
        y=df_country["vulnerable_pop"].iloc[-1] + 200000,
        s=country,
        fontsize=10,
        va='center',
        ha='right'
    )
# Remove legend
vuln_line_titles.legend_.remove()
```

# Change line thickness

```python
# Set country colours
country_colours = {
    "United Kingdom": "#1E2A16",
    "Argentina": "#D17E38",
    "Malawi": "#444443",
    "Nepal": "#3EA27F"
}

vuln_line_changed_thickness = sns.lineplot(
    data=vuln_4_country,
    x="year",
    y="vulnerable_pop",
    hue="country",
    palette=country_colours,
    # Change the thickness of the lines
    linewidth=3
)
# Set axis titles
vuln_line_changed_thickness.set_xlabel("Year")
vuln_line_changed_thickness.set_ylabel("Vulnerable Population")
# Set the gridlines for the x-axis
vuln_line_changed_thickness.yaxis.grid(True, linestyle='-', linewidth=0.5, color='lightgrey')
# Set y-axis title to be horizontal
vuln_line_changed_thickness.yaxis.label.set_rotation(0)
vuln_line_changed_thickness.yaxis.label.set_horizontalalignment('right')
vuln_line_changed_thickness.yaxis.label.set_verticalalignment('center')
vuln_line_changed_thickness.yaxis.set_label_coords(-0.1, 1.02)
# Remove scientific notation from y-axis
vuln_line_changed_thickness.ticklabel_format(axis='y', style='plain')
# Add labels to the end of the lines for each country
for country, df_country in vuln_4_country.groupby("country"):
    vuln_line_changed_thickness.text(
        x=df_country["year"].max(),
        y=df_country["vulnerable_pop"].iloc[-1] + 300000,
        s=country,
        fontsize=10,
        va='center',
        ha='right'
    )
# Remove legend
vuln_line_changed_thickness.legend_.remove()
```

# Add number suffixes (Thousands)

```python
# Set country colours
country_colours = {
    "United Kingdom": "#1E2A16",
    "Argentina": "#D17E38",
    "Malawi": "#444443",
    "Nepal": "#3EA27F"
}

vuln_line_thousands = sns.lineplot(
    data=vuln_4_country,
    x="year",
    y="vulnerable_pop",
    hue="country",
    palette=country_colours
)
# Set axis titles
vuln_line_thousands.set_xlabel("Year")
vuln_line_thousands.set_ylabel("Vulnerable Population")
# Set the gridlines for the x-axis
vuln_line_thousands.yaxis.grid(True, linestyle='-', linewidth=0.5, color='lightgrey')
# Remove scientific notation
vuln_line_thousands.ticklabel_format(axis='x', style='plain')
# Set y-axis title to be horizontal
vuln_line_thousands.yaxis.label.set_rotation(0)
vuln_line_thousands.yaxis.label.set_horizontalalignment('right')
vuln_line_thousands.yaxis.label.set_verticalalignment('center')
vuln_line_thousands.yaxis.set_label_coords(-0.1, 1.02)
# Remove scientific notation from y-axis and format ticks to be in thousands (for millions, divide by 1000000 and add "M" to the end, for billions, divide by 1000000000 and add "B" to the end)
vuln_line_thousands.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: "{:,}K".format(int(x/1000))))
# Add labels to the end of the lines for each country
for country, df_country in vuln_4_country.groupby("country"):
    vuln_line_thousands.text(
        x=df_country["year"].max(),
        y=df_country["vulnerable_pop"].iloc[-1] + 200000,
        s=country,
        fontsize=10,
        va='center',
        ha='right'
    )
# Remove legend
vuln_line_thousands.legend_.remove()
```

# Confidence intervals

```python
# Set country colours
from matplotlib.pyplot import fill_between


country_colours = {
    "United Kingdom": "#1E2A16",
    "Argentina": "#D17E38",
    "Malawi": "#444443",
    "Nepal": "#3EA27F"
}

vuln_line_confidence_intervals = sns.lineplot(
    data=vuln_4_country,
    x="year",
    y="vulnerable_pop",
    hue="country",
    palette=country_colours,
)
# Add confidence intervals
for country, df_country in vuln_4_country.groupby("country"):
    fill_between(
    df_country["year"],
    df_country["lower_confidence_limit"],
    df_country["upper_confidence_limit"],
    color=country_colours[country],
    alpha=0.2
)
# Set axis titles
vuln_line_confidence_intervals.set_xlabel("Year")
vuln_line_confidence_intervals.set_ylabel("Vulnerable Population")
# Set the gridlines for the x-axis
vuln_line_confidence_intervals.yaxis.grid(True, linestyle='-', linewidth=0.5, color='lightgrey')
# Set y-axis title to be horizontal
vuln_line_confidence_intervals.yaxis.label.set_rotation(0)
vuln_line_confidence_intervals.yaxis.label.set_horizontalalignment('right')
vuln_line_confidence_intervals.yaxis.label.set_verticalalignment('center')
vuln_line_confidence_intervals.yaxis.set_label_coords(-0.1, 1.02)
vuln_line_confidence_intervals.ticklabel_format(axis='y', style='plain')
# Add labels to the end of the lines for each country
for country, df_country in vuln_4_country.groupby("country"):
    vuln_line_confidence_intervals.text(
        x=df_country["year"].max(),
        y=df_country["vulnerable_pop"].iloc[-1] + 500000,
        s=country,
        fontsize=10,
        va='center',
        ha='right'
    )
# Remove legend
vuln_line_confidence_intervals.legend_.remove()
```

# Break the start of the y-axis

```python
# Create a new dataframe with the top 2 countries with the largest vulnerable population so there is a large empty space at the bottom of the chart
vuln_top_2 = vuln.loc[vuln["country"].isin(vuln.groupby("country")["vulnerable_pop"].sum().nlargest(2).index)]

vuln_axis_break = sns.lineplot(
    data=vuln_top_2,
    x="year",
    y="vulnerable_pop",
    hue="country"
)

# Break the y-axis so the chart begins at y = 5000000000 and ends at y = 30000000000
vuln_axis_break.set_ylim(50000000, 300000000)

# Add axis break symbol
kwargs = dict(
    transform=vuln_axis_break.transAxes,
    color='black',
    clip_on=False
)
# Draw two diagonal lines on the left axis
vuln_axis_break.plot(
    (-0.015, +0.015),
    (0.02, 0.06),
    **kwargs
)
vuln_axis_break.plot(
    (-0.015, +0.015),
    (0.06, 0.10),
    **kwargs
)
# Set the gridlines for the y-axis
vuln_axis_break.yaxis.grid(True, linestyle='-', linewidth=0.5, color='lightgrey')
# Set y-axis title to be horizontal
vuln_axis_break.yaxis.label.set_rotation(0)
vuln_axis_break.yaxis.label.set_horizontalalignment('right')
vuln_axis_break.yaxis.label.set_verticalalignment('center')
vuln_axis_break.yaxis.set_label_coords(-0.1, 1.04)
# Remove scientific notation from y-axis
vuln_axis_break.ticklabel_format(axis='y', style='plain')
# Add labels to the end of the lines for each country
for country, df_country in vuln_top_2.groupby("country"):
    vuln_axis_break.text(
        x=df_country["year"].max(),
        y=df_country["vulnerable_pop"].iloc[-1] + 10000000,
        s=country,
        fontsize=10,
        va='center',
        ha='right'
    )
# Remove legend
vuln_axis_break.legend_.remove()
```
